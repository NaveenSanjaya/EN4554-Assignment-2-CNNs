**Gradient Descent with a 1-D Function**$$f(w) = w^4 - w^3 - 12w^2 + 15w + 5$$

```python
import numpy as np
import matplotlib.pyplot as plt

def f(w):
    return w**4 - w**3 - 12*w**2 + 15*w + 5

def df(w):
    return 4*w**3 - 3*w**2 - 24*w + 15

w = np.linspace(-4, 4, 100)
plt.plot(w, f(w))

eta = 0.02
w0 = 0.2
plt.plot(w0, f(w0), 'ro')
w = w0
for i in range(10):
    w = w - eta * df(w) # Update w using gradient descent
    plt.plot(w, f(w), 'bo-')
    print(f"Iteration {i+1}: w = {w}, f(w) = {f(w)}")

plt.xlabel('w')
plt.ylabel('f(w)')
plt.title('Gradient Descent on f(w)')
plt.grid()
plt.show()
```

**Gradient Descent with PyTorch on the Same 1-D Function**

```python
import numpy as np
import torch
import matplotlib.pyplot as plt

def f(w):
    return w**4 - w**3 - 12*w**2 + 15*w + 5

w = torch.linspace(-4, 4, 100)
plt.plot(w.numpy(), f(w.numpy()))

eta = 0.02
w0 = 0.8
plt.plot(w0, f(w0), 'ro')
w0 = 0.2
w = torch.tensor(w0, requires_grad=True)
for i in range(10):
    # Forward pass
    loss = f(w)
    # Backward pass
    loss.backward()
    # Update w using gradient descent
    with torch.no_grad():
        w -= eta * w.grad
        w.grad.zero_()  # Reset gradients for the next iteration

    plt.plot(w.item(), f(w.item()), 'bo-')
    print(f"Iteration {i+1}: w = {w.item()}, f(w) = {f(w.item())}")

plt.xlabel('w')
plt.ylabel('f(w)')
plt.title('Gradient Descent on f(w) using PyTorch')
plt.grid()
plt.show()
```
